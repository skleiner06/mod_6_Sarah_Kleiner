## Reference sheet — Read before beginning

These are **small syntax fragments**, not complete programs. You must decide which ideas to use, where to place them, and how to combine them.

**Exception handling**
- `try:` begins the statements that might fail.
- `except ValueError:` catches an invalid numeric conversion or a deliberately raised `ValueError`.
- `raise ValueError("message")` deliberately signals an invalid value.
- `int(text)` converts whole-number text; `float(text)` converts decimal-number text.
- `else:` after `except` runs only when no exception occurs; `finally:` runs regardless of whether an exception occurs. Use these only if helpful.

**Sentinel-controlled repetition**
- A *sentinel* is a special input that ends repetition, such as `done`, `0`, or `-1`.
- `while answer != "done":` is an example of a continuation condition.
- Decide where to read the first value, where to check for the sentinel, and where to read the next value.
- A counter can be updated using `count += 1`; a total using `total += number`.
- For a numeric sentinel, decide how to handle invalid text without accidentally ending the loop.

**Nested loops**
- `for x in range(1, 4):` iterates over `1`, `2`, `3`; the stop value is excluded.
- `while x <= limit:` repeats as long as its condition is true.
- The *inner* loop completes before the *outer* loop advances.
- For `while` loops, identify which variables must change and which must be reset.
- `print(value, end=" ")` keeps output on the same line; `print()` begins a new line.
- `break` exits the nearest enclosing loop; `continue` skips to its next iteration. Use them thoughtfully.

**Boolean logic**
- `and` requires both operands to be true; `or` requires at least one; `not` reverses a Boolean.
- Precedence is `not`, then `and`, then `or`. Use parentheses to make your intent explicit.
- De Morgan's laws: `not (A and B)` equals `(not A) or (not B)`; `not (A or B)` equals `(not A) and (not B)`.
- Comparison operators: `==`, `!=`, `<`, `<=`, `>`, `>=`.
- `number % 2 == 0` tests whether an integer is even.

**Strings and lists — limited tools**
- `input("Prompt: ")` returns a string.
- `answer.strip().lower()` removes surrounding spaces and normalizes capitalization.
- `answer == "done"` compares a string with a sentinel.
- `for letter in word:` visits each character of a string, one at a time. You do not need to index the string.
- `len(word)` counts characters in a string; `len(items)` counts items in a list.
- `items[0]` accesses the first list item (indexing begins at zero).
- `for item in items:` visits each item in a list.
- If you choose to collect values in a list, start with `values = []`, then use `values.append(number)` to add a value. **Appending is not required in this lab.**

## Part 2 — Sentinel-controlled `while` loops (25 minutes)

### 2A. Sentinel tracing (5 minutes)

**Your tasks:**
- Predict the final **sum** and **count** for a loop that reads integers until the sentinel `0`.
- Use the input sequence `4`, `7`, `0`, `9` for your prediction.
- Record your prediction in comments **before coding**.
- Implement the sentinel-controlled loop and verify.
- Explain whether `0` and `9` are included.

**Test your work:** Try entering `0` first.

```python
# Count: 2, Sum: 11. Once 0 is entered, the loop will stop so it won't get to 9
# 0 isn't included because it is the sentinel

numbers = []
count = 0

while True:
  try:
    num = int(input("Enter an integer or 0 when your are done: "))

    if num == 0:
      break

    else:
      count += 1
      numbers.append(num)

  except ValueError:
    print("You must enter an integer.")

print("Count: ", count)
print("Sum: ", sum(numbers))
```

### 2B. Positive-number collector (14 minutes)

**Your tasks:**
- Continue requesting entries until the user types `done`, regardless of capitalization.
- Accept **positive whole numbers only**.
- Reject invalid text, zero, and negative numbers without ending the loop.
- Print the **count** and **sum** of accepted values at the end.
- Use `while`, `try`, `except ValueError`, and a deliberate `raise ValueError`.
- Use counters and totals; **no list is necessary**.

**Test your work:** `3`, `bad`, `-2`, `5`, `DONE` should produce count `2`, sum `8`.

```python
count = 0
total = 0

while True:
  try:
    answer = input("Enter a positive whole number, or done when your are finished: ")

    if answer.strip().lower() == 'done':
      break

    pos_num = int(answer)

    if pos_num < 1:
      raise ValueError("You must enter positive number.")

    else:
      count += 1
      total += pos_num

  except ValueError as e:
    if str(e) == "You must enter positive number.":
      print("You must enter a whole number.")

    else:
      print("You may only enter whole numbers.")
print("Count:", count)
print("Sum:", total)
```

### 2C. Sentinel with a special case (6 minutes)

**Your tasks:**

**Test your work:** `2`, `0`, `5`, `8`, `abc`, `-1` should produce count `2`.

```python
count = 0
while True:
  try:
    answer = int(input("Enter an even postive number, or -1 to stop: "))

    if answer == -1:
      break

    elif answer > 0 and answer % 2 == 0:
      count += 1

  except ValueError:
    print("You must enter an integer.")

print("Count: ", count)

# The location of the sentinel occurs even before it is checked to see if it an acceptable number or not, so that the loop breaks without any error messages.
```

```python
while True:
  try:
    row = int(input("Enter a number of rows between 1 and 4: "))

    if row >= 1 and row <= 4:
      break

    elif row < 1 or row > 4:
      raise ValueError("You must enter a number in range.")

  except ValueError as e:
    if str(e) == "You must enter a number in range.":
      print("You must enter a number in range.")

    else:
      print("You must enter an integer.")

row_count = 1

while row_count <= row:
  countdown = 3

  while countdown >= 1:
    print(countdown, end = ' ')
    countdown -= 1

  print()
  row_count += 1

# The countdown needs to be in the outer loop so that it resets each time. Otherwise, the second row would start from 0.
```

## Part 5 — Integrated challenge (15 minutes)

### 5A. Workshop attendance report (15 minutes)

**Your tasks:**
- Process **two workshops**.
- For each workshop, repeatedly request attendance counts until `done`.
- Accept **nonnegative integers only**.
- Reject invalid text and negative counts using `try`, `raise`, and `except ValueError` without ending the input loop.
- Classify a workshop as **Well attended** if **at least two valid entries are 5 or greater**; otherwise print **Needs outreach**.
- Use an **outer `for`** and an **inner sentinel-controlled `while`**.
- Use a Boolean condition for the classification and keep workshop counts separate.
- **No lists are needed**.
- **Optional extension:** Also require at least **three valid entries** for `Well attended`, and explain the change to your Boolean condition.

**Test your work:** Workshop 1: `4`, `bad`, `7`, `5`, `done`; workshop 2: `-1`, `5`, `done`.

```python
for workshop in range(1, 3):
  workshop_5_or_greater = 0
  total = 0

  print("Workshop", workshop)

  while True:
    try:
      attendance = input("Enter attendance count or done when finished: ").strip().lower()

      if attendance == 'done':
        break

      count = int(attendance)

      if count < 0:
        raise ValueError("Only nonnegative numbers allowed.")

      total += 1

      if count >= 5:
        workshop_5_or_greater += 1

    except ValueError as e:
      if str(e) == "Only nonnegative numbers allowed.":
          print("Only nonnegative numbers allowed.")
      else:
          print("You must enter an integer.")

  final = workshop_5_or_greater >= 2

  if final:
    print("Well attended.")

  else:
    print("Needs outreach.")

print()
```
